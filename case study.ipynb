```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
```

```python
data = df.copy()

data["Dt_Customer"] = pd.to_datetime(
    data["Dt_Customer"],
    dayfirst=True,
    errors="coerce"
)

latest_date = data["Dt_Customer"].max()

data["Customer_Tenure_Days"] = (
    latest_date - data["Dt_Customer"]
).dt.days
```

```python
data = data.drop(
    columns=[
        "ID",
        "Dt_Customer",
        "Z_CostContact",
        "Z_Revenue",
        "Response"
    ]
)
```

```python
print(data.head())
```

```python
print(data.shape)
```

```python
numeric_cols = data.select_dtypes(include=np.number).columns

for col in numeric_cols:
    data[col] = data[col].fillna(data[col].median())
```

```python
categorical_cols = data.select_dtypes(include="object").columns

for col in categorical_cols:
    data[col] = data[col].fillna(data[col].mode()[0])
```

```python
print(data.isnull().sum().sum())
```

```python
scaler = StandardScaler()

data[numeric_cols] = scaler.fit_transform(data[numeric_cols])

print(data.head())
```

```python
data = pd.get_dummies(
    data,
    columns=list(categorical_cols),
    dtype=int
)

print(data.head())
print("Data shape:", data.shape)
```

```python
X = data.values

print("Missing values:", data.isnull().sum().sum())
print("Data shape:", data.shape)
print("Feature matrix shape:", X.shape)
```

```python
kmeans_first = KMeans(n_clusters=4, random_state=42, n_init=10)

first_labels = kmeans_first.fit_predict(X)

print("Initial K-Means cluster sizes:")
print(pd.Series(first_labels).value_counts().sort_index())
```

```python
inertias = []
k_values = range(2, 8)

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X)
    inertias.append(model.inertia_)
```

```python
plt.plot(list(k_values), inertias, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

```python
silhouette_scores = []
k_values = range(2, 8)

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X)

    score = silhouette_score(X, labels)
    silhouette_scores.append(score)
```

```python
plt.plot(list(k_values), silhouette_scores, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

```python
k = 4

kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

kmeans_labels = kmeans.fit_predict(X)

results = df.copy()
results["KMeans_Cluster"] = kmeans_labels

print("Final K-Means cluster sizes:")
print(results["KMeans_Cluster"].value_counts().sort_index())

print("\nSilhouette Score:", round(silhouette_score(X, kmeans_labels), 3))
```

```python
summary_cols = [
    "Income", "MntWines", "MntMeatProducts", "MntFishProducts",
    "MntSweetProducts", "MntGoldProds", "NumWebPurchases",
    "NumCatalogPurchases", "NumStorePurchases", "NumWebVisitsMonth",
    "Recency"
]

print("Average customer values for each K-Means cluster:")
display(results.groupby("KMeans_Cluster")[summary_cols].mean().round(1))
```

```python
dbscan = DBSCAN(eps=4.0, min_samples=10)
```

```python
dbscan_labels = dbscan.fit_predict(X)

results["DBSCAN_Cluster"] = dbscan_labels

print("First 20 DBSCAN labels:")
print(dbscan_labels[:20])
```

```python
n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)

n_noise = np.sum(dbscan_labels == -1)

print("Number of clusters (excluding noise):", n_clusters)
print("Number of noise points:", n_noise)
print("Noise percentage:", round(100 * n_noise / len(dbscan_labels), 2), "%")
```

```python
print(pd.Series(dbscan_labels, name="DBSCAN_Cluster").value_counts().sort_index())
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

print("Explained variance ratio:", np.round(pca.explained_variance_ratio_, 3))

print("Total variance shown in 2D:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_labels, s=18)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA coloured by K-Means cluster")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=dbscan_labels, s=18)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA coloured by DBSCAN label (-1 = noise)")

plt.show()
```

```python
tsne = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate="auto",
    init="pca",
    random_state=42
)

X_tsne = tsne.fit_transform(X)

print("t-SNE shape:", X_tsne.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=kmeans_labels, s=18)

plt.xlabel("t-SNE dimension 1")
plt.ylabel("t-SNE dimension 2")
plt.title("t-SNE coloured by K-Means cluster")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=dbscan_labels, s=18)

plt.xlabel("t-SNE dimension 1")
plt.ylabel("t-SNE dimension 2")
plt.title("t-SNE coloured by DBSCAN label (-1 = noise)")

plt.show()
```
